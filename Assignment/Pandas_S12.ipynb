```python
import pandas as pd
```

```python
df = pd.read_csv("//content//IPL_Match_Results.csv")
df
```

```python
# Task 1.

team1 = df[["Team1", "Team1_Runs"]].rename(
    columns={"Team1": "Team", "Team1_Runs": "Runs"})

team2 = df[["Team2", "Team2_Runs"]].rename(
    columns={"Team2": "Team", "Team2_Runs": "Runs"})

all_runs = pd.concat([team1, team2])

total_runs = all_runs.groupby("Team")["Runs"].sum().reset_index()

print(total_runs)
```

```python
# Task 2.

import pandas as pd

df = pd.DataFrame({"city": ["Surat", "Surat", "Mumbai", "Mumbai", "Delhi", "Delhi"],
                  "restaurant": ["Domino's", "Domino's", "McDonald's", "McDonald's",
                   "KFC", "KFC"],
                    "order_amount": [500, 700, 800, 600, 450, 550]})

result = df.groupby(["city", "restaurant"])["order_amount"].agg("mean").reset_index()

print(result)
```

```python
# Task 3.
import pandas as pd

df = pd.DataFrame({"category": ["Mobile", "Mobile", "Laptop", "Laptop","Mobile", "Laptop"],
                    "region": ["West", "West", "West", "North", "North", "North"],
                    "sales": [50000, 70000, 80000, 90000, 60000, 75000],"units_sold": [10, 14, 8, 9, 12, 7]})

result = df.groupby( ["category", "region"]).agg(total_sales=("sales", "sum"),
    average_units_sold=("units_sold", "mean")).reset_index()

print(result)
```

```python
# Task 4.

df = pd.DataFrame({"region": ["West", "West", "North", "North", "South", "South"],"revenue": [50000, 70000, 80000, 60000, 45000, 55000]})

result = df.groupby("region").agg({"revenue": ["sum", "max", "min"]})

print(result)
```

```python
# Task 5.

import pandas as pd

df = pd.DataFrame({"artist": ["A", "B", "C", "D", "E", "F"],
    "genre": ["Pop", "Rock", "Pop", "Hip-Hop", "Rock", "Pop"],
    "streams": [500000, 300000, 700000, 400000, 600000, 800000]})

result = df.groupby("genre").agg(total_streams=("streams", "sum"),average_streams=("streams", "mean"))

result = result.sort_values("total_streams",ascending=False)

print(result)
```
