```python
import pandas as pd
```

```python
# Task 1.

movies = pd.DataFrame({'title': ['3 Idiots','Dangal','Sholay','Lagaan','Zindagi Na Milegi Dobara','Taare Zameen Par'],
                       'rating': [8.4, 8.3, 8.1, 8.1, 8.2, 8.3],
                       'release_year': [2009, 2016, 1975, 2001, 2011, 2007]})
print(movies)
print("==========")
print(movies.sort_values('rating', ascending=False))
```

```python
# Task 2.

import pandas as pd

ipl = pd.DataFrame({'team': ['CSK', 'MI', 'RCB', 'KKR', 'GT', 'RR'],'points': [17, 16, 14, 12, 18, 10],'net_run_rate': [0.65, 0.45, 0.32, 0.25, 0.75, -0.10]})
print(ipl)
print("==========")
print(ipl.sort_index(ascending=False))
```

```python
# Task 3.

import pandas as pd

youtube = pd.DataFrame({'video_title': ['Music Video','IPL Highlights','Comedy Show','Movie Trailer'],'views': [15000000, 12000000, 9500000, 8000000],
                        'likes': [1200000, 900000, 750000, 600000]})

youtube = youtube.reset_index(drop=True)

print(youtube)
```

```python
# Task 4.

df = pd.DataFrame({'restaurant_name': ['Domino’s', 'Biryani Blues', 'McDonald’s', 'Pizza Hut'],
                   'rating': [4.1, 4.5, 4.2, 4.0]}, index=[0, 1, 2, 3])

new_order = [1, 0, 2, 3]

df = df.reindex(new_order)
print(df)
```

```python
# Task 5.

df = pd.DataFrame({'product_name': ['Laptop', 'Headphones', 'Smartphone', 'Keyboard'],'price': [60000, 3000, 25000, 1500],
                   'discount': [20, 40, 15, 30]})

df = df.sort_values(by='discount', ascending=False).reset_index(drop=True)

print(df)
```
