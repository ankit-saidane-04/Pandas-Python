```python
# Task 1.

import pandas as pd

dates = pd.Series(["15-01-2026","23-03-2026","07-05-2026","19-08-2026","26-09-2026"])

dates = pd.to_datetime(dates, format="%d-%m-%Y")

print(dates)
```

```python
# Task 2.

import pandas as pd

df = pd.DataFrame({"movie": ["Movie A", "Movie B", "Movie C"],"release_date": ["15-01-2026", "23-03-2025", "07-05-2024"]})

df["release_date"] = pd.to_datetime(df["release_date"],format="%d-%m-%Y")

df["year"] = df["release_date"].dt.year
df["month"] = df["release_date"].dt.month
df["day"] = df["release_date"].dt.day

print(df)
```

```python
# Task 3.

import pandas as pd

dates = pd.date_range(start=pd.Timestamp.today().normalize(),periods=7)

print(dates)
```

```python
# Task 4.

import pandas as pd

df = pd.DataFrame({"user_id": [101, 102, 103],"start_date": ["01-01-2026", "15-03-2026", "10-07-2026"]})

df["start_date"] = pd.to_datetime(df["start_date"],format="%d-%m-%Y")

today = pd.Timestamp.today().normalize()

df["tenure_days"] = (today - df["start_date"]).dt.days

print(df)
```
