```python
# Task 1.

import pandas as pd

jan = pd.DataFrame({"Order_ID": [101, 102, 103],"Customer": ["Amit", "Rahul", "Priya"],"Amount": [450, 300, 550]})

feb = pd.DataFrame({"Order_ID": [104, 105, 106],"Customer": ["Neha", "Raj", "Kiran"],"Amount": [400, 650, 350]})

orders = pd.concat([jan, feb], ignore_index=True)

print(orders)
```

```python
# Task 2.

products = pd.DataFrame({"Product_ID": [101, 102, 103, 104], "Product": ["Laptop", "Mobile", "Headphones", "Watch"],"Price": [55000, 25000, 3000, 5000]})

ratings = pd.DataFrame({"Rating": [4.5, 4.2, 4.0, 4.4]})

result = pd.concat([products, ratings], axis=1)

print(result)
```

```python
# Task 4.

orders = pd.DataFrame({"restaurant": ["KFC", "KFC", "KFC","Dominos", "Dominos", "Dominos","McDonalds", "McDonalds", "McDonalds"],
                      "month": ["January", "February", "March",
                                 "January", "February", "March",
                                "January", "February", "March"],
    "total_amount": [500, 600, 550,400, 450, 500,350, 400, 450]})

result = pd.pivot_table(orders,
                        index="restaurant",
                        columns="month",
                        values="total_amount",
                        aggfunc="mean")

print(result)
```
