```python
ipl_df =df
conditions =[ipl_df["Win_By_Runs"] > 0, ipl_df["Win_By_Wickets"] > 0]

choices = ["Runs","Wickets"]

ipl_df["match_margin_type"] = np.select(conditions,choices,default="Tie/No Result")

ipl_df.head()
```

```python
team_city = pd.DataFrame({"Team": ["GT", "RCB", "MI", "CSK", "KKR", "DC", "SRH", "RR"],
                          "Home_City": ["Ahmedabad","Bengaluru","Mumbai","Chennai","Kolkata","Delhi","Hyderabad", "Jaipur"]})

print(team_city)

#Team 1
ipl_df = ipl_df.merge(team_city,left_on="Team1",right_on="Team",how="left")

ipl_df = ipl_df.rename(columns={"Home_City": "Team1_Home_City"})

ipl_df = ipl_df.drop(columns=["Team"])

#Team 2
ipl_df = ipl_df.merge(team_city,
    left_on="Team2",
    right_on="Team",
    how="left")

ipl_df = ipl_df.rename(columns={"Home_City": "Team2_Home_City"})

ipl_df = ipl_df.drop(columns=["Team"])
```

```python
print(ipl_df[["Team1", "Team1_Home_City","Team2","Team2_Home_City"]].head(10))
```

```python
# Create pivot table: total wins by each team per season

wins_pivot = pd.pivot_table(ipl_df,index="Winner",columns="Season",values="Match_ID",aggfunc="count",fill_value=0)

print(wins_pivot)
```

```python
# Convert pivot table into a normal DataFrame

top_3 = (wins_pivot.reset_index().melt(id_vars="Winner", var_name="Season", value_name="Wins")
          .sort_values("Wins", ascending=False).head(3))

print(top_3)
```
